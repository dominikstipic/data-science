```python
import requests
from llm import deepseek_input_prompt


url = "https://www.imc-tm.com/about-imc/privacy"
headers = {
        "User-Agent": "Mozilla/5.0 (X11; Ubuntu; Linux x86_64; rv:149.0) Gecko/20100101 Firefox/149.0",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8"
    }
response = requests.get(url, headers=headers)
data = response.text

out = deepseek_input_prompt(
    "Create a dataframe table which shows the personal data which company collects", 
    data)
```

```python
import re
import json

raw_text = """
**Personal Details**
*   Full Name
*   Date of Birth
*   Gender / Title
*   Home Address
*   Personal Email Address
*   Personal Phone/Mobile Number
*   Government ID Numbers (e.g., Social Security Number, Passport Number)
*   Nationality & Citizenship
*   Marital Status
*   Dependent Information
*   Emergency Contact Details
*   Signature (physical or digital)
*   Biometric Data (e.g., fingerprint, facial recognition for access)
*   Photograph or Video Image
*   Voice Recording (from customer service calls)
*   Racial or Ethnic Origin Data
*   Religious or Philosophical Beliefs
*   Trade Union Membership
*   Health & Medical Information
*   Genetic Data
*   Sexual Orientation
*   Financial Account Numbers
*   Payment Card Details
*   Credit History & Score
*   Tax Identification Number
*   Employment History
*   Educational Background
*   Professional Licenses & Certifications
*   Social Media Profile Handles/URLs
*   IP Address (when linked to an account)
*   Device Identifiers (e.g., IMEI, MAC address)
*   Location Data (precise GPS)
*   Web Browsing History & App Usage
*   Purchase History & Transaction Records
*   Communication Preferences
*   Survey & Feedback Responses
*   Hobbies & Interests (inferred or provided)
*   Family & Household Information
*   Political Opinions or Affiliation
"""

# Step 1: Extract bullet items
items = re.findall(r"\*\s+(.*)", raw_text)

# Step 2: Categorize fields
categories = {
    "identity": [],
    "contact": [],
    "sensitive": [],
    "financial": [],
    "technical": [],
    "behavioral": [],
    "professional": [],
    "other": []
}

for item in items:
    lower = item.lower()

    if any(k in lower for k in ["name", "birth", "gender", "nationality", "marital"]):
        categories["identity"].append(item)

    elif any(k in lower for k in ["email", "phone", "address", "contact"]):
        categories["contact"].append(item)

    elif any(k in lower for k in [
        "racial", "religious", "health", "genetic", "sexual", "biometric", "union"
    ]):
        categories["sensitive"].append(item)

    elif any(k in lower for k in [
        "financial", "card", "credit", "tax", "payment"
    ]):
        categories["financial"].append(item)

    elif any(k in lower for k in [
        "ip", "device", "location"
    ]):
        categories["technical"].append(item)

    elif any(k in lower for k in [
        "history", "preferences", "survey", "hobbies", "purchase", "browsing"
    ]):
        categories["behavioral"].append(item)

    elif any(k in lower for k in [
        "employment", "education", "license", "certification"
    ]):
        categories["professional"].append(item)

    else:
        categories["other"].append(item)

# Step 3: Output structured JSON
print(json.dumps(categories, indent=2))
```
