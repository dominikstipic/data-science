```python
from bs4 import BeautifulSoup
import requests
from dataclasses import dataclass

url = "https://optiver.com/news/?numberposts=6&paged=3&tax_query[topic]=0"
html = requests.get(url).text

bs = BeautifulSoup(html, "html.parser")
lis = bs.body.find("div", class_="result items items-viewmode-grid").find_all("li")


img_link = lis[0].header.img["src"]
footer = lis[0].footer.text.strip()
head_text = lis[0].main.span.text
news_link = lis[0].h2.a["href"]
news_link
```

```python
@dataclass
class OptiverArticle:
    _img_link = None
    _footer = None
    _head_text = None
    _news_link = None

    def __init__(self, img_link, footer, head_text, news_link):
        self._img_link = img_link
        self._footer = footer
        self._head_text = head_text
        self._news_link = news_link

    def __repr__(self):
        return self._news_link

def optiver_generator():
    lis = bs.body.find("div", class_="result items items-viewmode-grid").find_all("li")
    for li in lis:
        import pdb; pdb.set_trace()
        if "img" in li.__dict__:
            img_link = li.header.img["src"]
        else:
            img_link = None
        if "footer" in li.__dict__:
            footer = li.footer.text.strip()
        else:
            footer = None
        if "main" in li.__dict__:
            head_text = li.main.span.text
        else:
            head_text = None
        if "h2" in li.__dict__:
            news_link = li.h2.a["href"]
        else:
            news_link = ""
        oa = OptiverArticle(img_link, footer, head_text, news_link)
        yield oa

url = "https://optiver.com/news/?numberposts=6&paged=3&tax_query[topic]=0&paged=2"
html = requests.get(url).text
bs = BeautifulSoup(html, "html.parser")
lis = bs.body.find("div", class_="result items items-viewmode-grid").find_all("li")


list(optiver_generator())
```
